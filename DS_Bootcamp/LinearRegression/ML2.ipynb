```python
import pandas as pd
import numpy as np
import random
```

```python
train_part = pd.read_json('data/train.json')
test_part = pd.read_json('data/test.json')
```

```python
train_part.shape
```

```python
train_qun_99 = train_part['price'].quantile(0.99)
train_qun_01 =  train_part['price'].quantile(0.01)
train_part = train_part[(train_part['price'] < train_qun_99) & (train_part['price'] > train_qun_01)]

test_qun_99 = test_part['price'].quantile(0.99)
test_qun_01 =  test_part['price'].quantile(0.01)
test_part = test_part[(test_part['price'] < test_qun_99) & (test_part['price'] > test_qun_01)]
```

```python
train_part['interest_level'] = train_part['interest_level'].astype('category').cat.rename_categories(new_categories={'high': 2, 'medium': 1, 'low': 0})
```

```python
train_part
```

```python
result_array = []
for index, row in train_part.iterrows():
    for word in row['features']:
        word = word.replace(' ', '').replace('[', '').replace(']', '').replace("'", "").replace('"', '') 
        result_array.append(word)

nunique_rows = pd.Series(result_array).nunique()
nunique_rows
```

```python
result_array = []
for index, row in test_part.iterrows():
    for word in row['features']:
        word = word.replace(' ', '').replace('[', '').replace(']', '').replace("'", "").replace('"', '') 
        result_array.append(word)

nunique_rows = pd.Series(result_array).nunique()
nunique_rows
```

```python
from collections import Counter

most_common_features = Counter(result_array).most_common(20)
most_common_features
```

```python
top_features = []
for tuple in most_common_features:
    feature, count = tuple
    top_features.append(feature)

for feature in top_features:
    train_part[feature] = 0.0
    for i in range(train_part.shape[0]):
        if isinstance(train_part['features'].iloc[i], list) and feature in train_part['features'].iloc[i]:
            train_part[feature].iloc[i] = 1.0 
        else:
            train_part[feature].iloc[i] = 0.0   
```

```python
for feature in top_features:
    mask = test_part['features'].apply(lambda x: feature in x if isinstance(x, list) else False)
    test_part[feature] = mask.astype(float)
```

```python
train_part.fillna({'bathrooms': train_part['bathrooms'].mode()[0]}, inplace=True)
train_part.fillna({'bedrooms': train_part['bedrooms'].mode()[0]}, inplace=True)
```

```python
X_train = train_part[['bathrooms', 'bedrooms'] + top_features].values
y_train = train_part['price'].values
X_test = test_part[['bathrooms', 'bedrooms'] + top_features].values
y_test = test_part['price'].values

data = pd.DataFrame()
```

```python
np.random.seed(21)
class Lin_Reg_GD:
    def __init__(self, num_objects, num_features, features, y):
        self.num_objects = num_objects
        self.num_features = num_features + 1
        self.features = np.column_stack([np.ones(len(features)), features])
        self.y = y
        np.random.seed(21)
        self.W = np.random.randn(self.num_features)
       
    def MSE(self, y, predictions):
        return (np.mean((y - predictions) ** 2))
    
    def R2(self, y, predictions):
        return 1 - np.sum((y - predictions) ** 2)/np.sum((y - np.mean(y)) ** 2)

    def MAE(self, y, predictions):
        return (np.mean(np.abs(y - predictions)))

    def derivative_MSE(self, predictions):
        der_MSE = -2/(self.num_objects) * self.features.T @ (self.y - predictions)
        return der_MSE
        
    def update_params(self, learning_rate):
        predictions = self.predictions()
        self.W -= learning_rate * self.derivative_MSE(predictions)
        return self.W
    
    def predictions(self):
        predictions = self.features @ self.W 
        return predictions
    
    def predict(self, X_new):
        X_new_with_bias = np.column_stack([np.ones(len(X_new)), X_new])
        return X_new_with_bias @ self.W

class SGD(Lin_Reg_GD):
    def __init__(self, num_objects, num_features, features, y):
            super().__init__(num_objects, num_features, features, y) 

    def Lasso(self, mse, alpha):
        lasso =  mse + alpha * np.abs(self.W) 
        return lasso
    
    def Ridge(self, mse, alpha):
        ridge = mse + alpha * (self.W ** 2)
        return ridge
    
    def ElasticNet(self, mse, alpha, proportion):
        l1_n_l2 = mse + alpha * (proportion * (self.W)**2 + (1 - proportion) * np.abs(self.W))
        return l1_n_l2
            
    def update_random_param(self, learning_rate,  regularization='SGD', alpha=0.1, proportion=0.5):
        ind = np.random.randint(0, self.num_objects)
        x_ind = self.features[ind]
        y_ind = self.y[ind]
        prediction = x_ind @ self.W
        der_MSE = -2 * x_ind * (y_ind - prediction)

        if regularization == 'Ridge':
            loss = self.Ridge(der_MSE, alpha)
        elif regularization == 'Lasso':
            loss = self.Lasso(der_MSE, alpha)
        elif regularization == 'Elastic':
            loss = self.ElasticNet(der_MSE, alpha, proportion)
        else:
            loss = der_MSE
        
        self.W -= learning_rate * loss
    
```

```python
def studing_GD(iterations):
    LinReg = Lin_Reg_GD(X_train.shape[0], X_train.shape[1], X_train, y_train)

    for i in range(iterations): 
        LinReg.update_params(learning_rate=0.1)
    
    predictions = LinReg.predict(X_test)
    print(f'RMSE: {LinReg.MSE(y_test, predictions) ** 0.5}')
    print(f'R2: {LinReg.R2(y_test, predictions)}')
    print(f'MAE: {LinReg.MAE(y_test, predictions)}')

studing_GD(iterations=500)
```

```python
np.random.seed(21)
data = pd.DataFrame()

def studing_SGD(iterations, regularization='SGD', alpha=1, proportion=0.5, learning_rate=0.01, X_train=X_train, y_train=y_train, X_test=X_test , y_test=y_test):
    np.random.seed(21)
    LR_SGD = SGD(X_train.shape[0], X_train.shape[1], X_train, y_train)

    for i in range(iterations): 
        LR_SGD.update_random_param(learning_rate=learning_rate, regularization=regularization, alpha=alpha, proportion=proportion)
    
    predictions = LR_SGD.predict(X_test)

    global data
    new_model = {'Model': f'self-created_{regularization}', 
                  'RMSE': LR_SGD.MSE(y_test, predictions) ** 0.5, 
                  'R2': LR_SGD.R2(y_test, predictions), 
                  'MAE': LR_SGD.MAE(y_test, predictions)}
    data = pd.concat([data, pd.DataFrame([new_model])], ignore_index=True)

    print(f'RMSE: {LR_SGD.MSE(y_test, predictions) ** 0.5}')
    print(f'R2: {LR_SGD.R2(y_test, predictions)}')
    print(f'MAE: {LR_SGD.MAE(y_test, predictions)}')

studing_SGD(iterations=1000, learning_rate=0.001)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error 

def sklearnResults(X_train=X_train, X_test=X_test):
    model = LinearRegression()
    model.fit(X_train, y_train)

    global data
    y_pred = model.predict(X_test)
    new_model = {'Model': f'sklearn_LR', 
                    'RMSE': mean_squared_error(y_test, y_pred)** 0.5, 
                    'R2': r2_score(y_test, y_pred), 
                    'MAE': mean_absolute_error(y_test, y_pred)}
    data = pd.concat([data, pd.DataFrame([new_model])], ignore_index=True)
    print(new_model)

sklearnResults()
```

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

def regularization(model, X_train=X_train, X_test=X_test, **kwargs):
    reg_model = model(**kwargs)
    reg_model.fit(X_train, y_train)

    y_pred = reg_model.predict(X_test)

    global data
    model_name = str(model).replace('>', '').replace("'", "").split('.')[-1]
    new_model = {'Model': f'sklearn_{model_name}', 
                  'RMSE': mean_squared_error(y_test, y_pred)** 0.5, 
                  'R2': r2_score(y_test, y_pred), 
                  'MAE': mean_absolute_error(y_test, y_pred)}
    data = pd.concat([data, pd.DataFrame([new_model])], ignore_index=True)
    print(new_model)
```

```python
def Scaler(type, X):
    if type == 'Standard':
        X_scaled = (X - np.mean(X, axis=0)) / (np.std(X, axis=0) +  1e-8)
        return X_scaled
    elif type == 'MinMax':
        X_scaled = (X - np.min(X, axis=0)) / (np.max(X, axis=0) - np.min(X, axis=0) +  1e-8)
        return X_scaled
```

```python
X_train_standard = Scaler(type='Standard', X=X_train)
X_test_standard = Scaler(type='Standard', X=X_test)
studing_SGD(iterations=1000, learning_rate=0.001, X_train=X_train_standard, X_test=X_test_standard)
```
