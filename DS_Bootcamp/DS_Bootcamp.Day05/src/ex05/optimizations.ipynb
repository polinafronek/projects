## Calculation (fines/refund*year) by various methods

```python
%%timeit 

def calculate():
    calculated_data = []
    for i in range(0, len(fines)):
        calculated_row = fines.iloc[i, 2] / fines.iloc[i, 1] * fines.iloc[i, 5]
        calculated_data.append(calculated_row)
    fines["loop"] = calculated_data

calculate()
```

```python
%%timeit

for index, row in fines.iterrows():
    fines.loc[index, "Iterrows"] = row["Fines"] / row["Refund"] * row["Year"]
```

```python
%%timeit

fines["Apply"] = fines.apply(lambda x: x["Fines"] / x["Refund"] * x["Year"], axis=1)
```

```python
%%timeit

fines["Series"] = fines["Fines"] / fines["Refund"] * fines["Year"]
```

```python
%%timeit

fines["Values"] = fines["Fines"].values / fines["Refund"].values * fines["Year"].values
```

## Downcasting
* run df.info(memory_usage=’deep’), pay attention to the Dtype and the memory usage
* make a copy() of your initial dataframe into another dataframe optimized
* downcast from float64 to float32 for all the columns
* downcast from int64 to the smallest numerical dtype possible
* run info(memory_usage='deep') for your new dataframe, pay attention to the Dtype and the memory usage

```python
fines.info(memory_usage="deep")
```

```python
optimized = fines.copy()

optimized 
```

```python
for column in optimized.columns:
    if optimized[column].dtype == "float64":
        optimized[column] = optimized[column].astype("float32")
    elif optimized[column].dtype == "int64":
        optimized[column] = optimized[column].astype("int8")

optimized.info(memory_usage='deep')
```

## Change the object type columns to the type category

```python
for column in optimized.columns:
    if optimized[column].dtype == "object":
        optimized[column] = optimized[column].astype("category")

optimized.info(memory_usage='deep')
```
